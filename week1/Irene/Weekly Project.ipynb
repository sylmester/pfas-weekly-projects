```python
import cv2
import numpy as np
import imutils
from matplotlib import pyplot as plt
```

```python
path = "appletree.jpg"
```

```python
bgr_img = cv2.imread(path)
b,g,r = cv2.split(bgr_img)       # get b,g,r
image = cv2.merge([r,g,b])
plt.imshow(image)
```

```python
red_up_lim=np.array([255,70,130], dtype="uint8")
red_low_lim=np.array([80,20,10], dtype="uint8")
mask = cv2.inRange(image, red_low_lim, red_up_lim)
output = cv2.bitwise_and(image, image, mask=mask)
plt.imshow(output)
```

```python
gray = cv2.cvtColor(output, cv2.COLOR_RGB2GRAY)
plt.imshow(gray, cmap = 'gray')
```

```python
threshold = 10
threshold_value = 255

thresh = cv2.threshold(gray, threshold, threshold_value, cv2.THRESH_BINARY)[1]
plt.imshow(thresh, cmap='gray')
```

## 4) Can you segment the leaves?

To get this part done, we should proceed in a similar way as with the segmentation of the apples. This time we need to focus on the color green, even though at first sight you would probably think that red is not needed, as there are some leaves brighter so the upper limit may need some red to get closer to the white. On the other hand, there are some parts of the apples appearing but just slightly.

```python
green_up_lim=np.array([100,255,35], dtype="uint8")
green_low_lim=np.array([20,52,0], dtype="uint8")
mask = cv2.inRange(image, green_low_lim, green_up_lim)
output = cv2.bitwise_and(image, image, mask=mask)
plt.imshow(output)
```
